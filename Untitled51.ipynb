```python
from tkinter import *

window = Tk()
window.title("To-Do List")
window.geometry("500x400")

label1 = Label(window,text="Enter Task :",fg="red",font=("Arial",14))
label1.grid(row=0,column=0,padx=10,pady=10)


textbox = Entry( window, width=30,font=("Arial",14))
textbox.grid(row=0,column=1,padx=10,pady=10)

tasklist = Listbox(window, width=40,height=12,font=("Arial",12))

tasklist.grid(row=2,column=0,columnspan=3,padx=20,pady=20)

def addTask():

    task = textbox.get()

    if task != "":
        tasklist.insert(END, task)
        textbox.delete(0, END)

def deleteTask():

    selected = tasklist.curselection()

    if selected:
        tasklist.delete(selected)

def completeTask():

    selected = tasklist.curselection()

    if selected:

        task = tasklist.get(selected)

        tasklist.delete(selected)

        tasklist.insert(selected, "✓ " + task)
        
button1 = Button( window,text="Add Task",command=addTask,font=("Arial",12))
button1.grid(row=1,column=0,pady=10)
button2 = Button(window,text="Complete",command=completeTask,font=("Arial",12))
button2.grid(row=1,column=1,pady=10)
button3 = Button(window,text="Delete",command=deleteTask, font=("Arial",12))
button3.grid(row=1,column=2,pady=10)

window.mainloop()
```
