```python
n1 = float(input("Enter the 1st number "))
n2 = float(input("Enter the 2nd number "))
print("intializing calculator...")

operation = input("Enter your opertaion: ")

if operation == "add" or operation == "addition":
    Ans= n1 + n2 ;print("Result:", Ans)

elif operation == "sub" or operation =="subtraction":
    Ans = n1 - n2;print("Result:", Ans)

elif operation == "multi" or operation == "multiplication":
    Ans = n1 * n2; print("Result:", Ans)

elif operation == "div" or operation == "division":
    if n2 != 0:
        Ans = n1 / n2; print("Result:", Ans)
    else:
        print("Error: zero division")

else:
    print("Invalid operation...")
```
